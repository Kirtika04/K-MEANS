```python
def euclideandistance(x,y):
    return np.sqrt(np.sum((x-y)**2))
```

```python
def kmeans(data,k=3):
    X=data.values
    #centroids=X[:k].copy() #to implement first k instances as the initial centroids
    centroids=[]
    indexes=[]
    for i in random.sample(range(0,100),k):
        centroids.append(X[i])
        indexes.append(i)
    print("INITIAL VALUES TUPLE NO.:",indexes)
    centroids=np.array(centroids)
    for i in range(15):
        clusters=[[] for i in range(k)]
        labels=[]
        for point in X:
            distances=[]
            for centroid in centroids:
                distances.append(euclideandistance(point,centroid))
            clusters[distances.index(min(distances))].append(point)
            labels.append(distances.index(min(distances)))
        newcentroids=[]
        for cluster in clusters:
            newcentroids.append(np.mean(cluster,axis=0))
        newcentroids=np.array(newcentroids)
        if np.allclose(centroids,newcentroids):
            print("CONVERGENCE HAPPENED!!!!!")
            break
        centroids=newcentroids
        print("ITERATION",i+1)
        for cluster,i in zip(clusters,range(1,k+1)):
            print("CLUSTER",i,":",len(cluster))
    return clusters,centroids,labels   
```

```python
clusters,centroids,labels=kmeans(data,3)
#print(labels)
```

```python
labels=np.array(labels)
```

```python
for i,j in enumerate(['yellow','blue','green']):
    p=(labels==i)
    plt.scatter(X[p,2],X[p,3],c=j,s=50,label="C"+str(i+1)) #ALL ROWS, COLUMN NO. c colour and s size 
plt.scatter(centroids[:,2],centroids[:,3],marker="X",s=200,c="red",label="Centroids")
plt.xlabel("PetalLengthCm")
plt.ylabel("PetalWidthCm")
plt.title("K MEANS CLUSTERING")
plt.legend()
plt.show()
```
